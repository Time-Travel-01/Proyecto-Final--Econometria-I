### **Identificación de problemas en el modelo**

#### **1. Multicolinealidad (VIFs)**

```python
# Factor de Inflación de la Varianza (VIF)
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Variable dependiente (Gasto)
gasto = datos['gasto_mon']

# Variables explicativas
X = data_frame1[[
    "Alimentos",
    "Transporte",
    "Ingresos",
    "Personales",
    "Limpieza",
    "Educación",
    "Vivienda",
    "Salud",
    "Integrantes",
    "Nivel_Educativo",   # categórica
    "Estrato",           # categórica
    "Zona",              # categórica
    "Sexo"               # categórica
]]

# Convertir las variables categóricas a variables dummy
X = pd.get_dummies(X, drop_first=True, dtype=float)

# Agregar el intercepto
X = sm.add_constant(X)

# Asegurarse de que todo sea numérico
X = X.astype(float)

# Se calcula el VIF para cada columna
vif = pd.DataFrame({
    "Variable": X.columns,
    "VIF": [variance_inflation_factor(X.values, i)
            for i in range(X.shape[1])]
})

# Se excluye la constante del análisis dado que presenta VIF alto,
# lo cual en realidad, no es de nuestro interés por lo cual,
# no es muy significativo
vif = vif[vif["Variable"] != "const"]

# Ordenar de mayor a menor VIF
vif = vif.sort_values("VIF", ascending=False).reset_index(drop=True)
# Resultado final
vif
```

Los valores obtenidos de los VIFs nos indican una baja o moderada multicolinealidad entre los predictores, por lo cual **se descarta este problema del modelo**.

#### **2. Normalidad de residuos (Histograma, Puebas de Jarque-Bera y de Shapiro-Wilk)**

```python
# Residuos del modelo
residuos = modelo2_df1.resid

# Histograma de los residuos
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(residuos, bins=100, edgecolor="black")
plt.xlabel("Residuos")
plt.ylabel("Frecuencia")
plt.title("Histograma de los residuos")
plt.show()

# Gráfico Q-Q
import statsmodels.api as sm

sm.qqplot(residuos, line="45", fit=True)
plt.title("Q-Q Plot de los residuos")
plt.show()

# Prueba de Jarque-Bera
from scipy.stats import jarque_bera

jb_stat, jb_pvalue = jarque_bera(residuos)

print("Prueba de Jarque-Bera")
print(f"Estadístico JB: {jb_stat:.4f}")
print(f"p-value: {jb_pvalue:.6f}")

# Prueba de Shapiro-Wilk
# NOTA: es recomendable solo para muestras pequeñas o medianas,
# por lo cual, no es funcional en nuestro caso.
from scipy.stats import shapiro

shapiro_stat, shapiro_pvalue = shapiro(residuos)

print("\nPrueba de Shapiro-Wilk")
print(f"Estadístico W: {shapiro_stat:.4f}")
print(f"p-value: {shapiro_pvalue:.6f}")

# ------------ Interpretación de las Pruebas ------------
alpha = 0.05

print("\nInterpretación (Jarque-Bera):")
if jb_pvalue > alpha:
    print("No se rechaza H0: los residuos son compatibles con una distribución normal.")
else:
    print("Se rechaza H0: los residuos no siguen una distribución normal.")

print("\nInterpretación (Shapiro-Wilk):")
if shapiro_pvalue > alpha:
    print("No se rechaza H0: los residuos son compatibles con una distribución normal.")
else:
    print("Se rechaza H0: los residuos no siguen una distribución normal.")
```

#### **3. Heterocedasticidad (Gráficos y Prueba de Breusch-Pagan)**

```python
# Extraer residuos y valores ajustados del modelo
residuos # ya los tenemos del chunk anterior
ajustados = modelo2_df1.fittedvalues

# Gráfico residuos vs valores ajustados
plt.figure(figsize=(8, 5))
plt.scatter(ajustados, residuos, alpha=0.5)
plt.axhline(y=0, linestyle="--")
plt.xlabel("Valores ajustados")
plt.ylabel("Residuos")
plt.title("Residuos vs Valores Ajustados")
plt.show()

# Prueba de Breusch-Pagan
from statsmodels.stats.diagnostic import het_breuschpagan

# La matriz de diseño (incluye la constante y las dummies
# generadas automáticamente por C(...))
X = modelo2_df1.model.exog

bp_test = het_breuschpagan(residuos, X)

bp_labels = [
    "LM Statistic",
    "LM p-value",
    "F Statistic",
    "F p-value"
]

print("Prueba de Breusch-Pagan")
for nombre, valor in zip(bp_labels, bp_test):
    print(f"{nombre}: {valor:.6f}")

# 5. Interpretación de la Prueba
alpha = 0.05

print("\nInterpretación (Breusch-Pagan):")
if bp_test[1] > alpha:
    print("No se rechaza H0: existe homocedasticidad.")
else:
    print("Se rechaza H0: existe heterocedasticidad.")
```

#### **4. Especificación (Prueba de Ramsey RESET)**

```python
# Juego de hipótesis
# H0: El modelo está correctamente especificado.
# H1: El modelo está mal especificado (faltan variables,
#     forma funcional incorrecta o términos no lineales).

# Importar la función
from statsmodels.stats.diagnostic import linear_reset

# Aplicar la prueba al modelo estimado
reset_test = linear_reset(modelo2_df1, power=2, use_f=True)

# Resultados obtenidos
print("Prueba de Ramsey RESET")
print(f"Estadístico F : {reset_test.fvalue:.6f}")
print(f"p-value       : {reset_test.pvalue:.6f}")

# Interpretación de la prueba
alpha = 0.05

print("\nInterpretación:")
if reset_test.pvalue > alpha:
    print("No se rechaza H0: el modelo parece estar correctamente especificado.")
else:
    print("Se rechaza H0: existe evidencia de mala especificación del modelo.")
```

### **IDEA 2: Identificación de Valores Influyentes (distancia de Cook, leverage y residuos studentizados), Outliers (winsorización)**

#### **Distancia de Cook**
La distancia de Cook es una medida estadística que se utiliza para **identificar observaciones influyentes** en un conjunto de datos al ajustar un modelo de regresión lineal.

En términos más sencillos, ayuda a comprender **cuánto cambiaría el modelo si se eliminara un dato específico**. *Cuanto mayor sea la distancia de Cook, mayor será la influencia de dicho dato*.

```python
influencia = modelo2_df1.get_influence()
dis_cook = influencia.cooks_distance[0]

umbral = 4 / len(data_frame1) # umbral un tanto estricto, faltaría checar si es correcto tomarlo de esta forma
outliers = np.where(dis_cook > umbral)[0]

print("Número de observaciones influyentes:", len(outliers))
#print("Índice de observaciones influyentes:", outliers)
# DataFrame con las filas que contienen observaciones influyentes
df = data_frame1.loc[outliers]

# Exportar este fragmento de DataFrame a excel
df.to_excel("outliers.xlsx")
```

###### **1. Multicolinealidad**

```python
# Factor de Inflación de la Varianza (VIF)
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Variable dependiente (Gasto)
gasto = data_frame1['log_gasto']

# Variables explicativas
X = data_frame1[[
    "log_alimentos",
    "log_transporte",
    "log_ingresos",
    "log_personales",
    "log_limpieza",
    "log_educacion",
    "log_vivienda",
    "log_salud",
    "Integrantes",
    "Nivel_Educativo",   # categórica
    "Estrato",           # categórica
    "Zona",              # categórica
]]

# Convertir las variables categóricas a variables dummy
X = pd.get_dummies(X, drop_first=True, dtype=float)

# Agregar el intercepto
X = sm.add_constant(X)

# Asegurarse de que todo sea numérico
X = X.astype(float)

# Se calcula el VIF para cada columna
vif = pd.DataFrame({
    "Variable": X.columns,
    "VIF": [variance_inflation_factor(X.values, i)
            for i in range(X.shape[1])]
})

# Se excluye la constante del análisis dado que presenta VIF alto,
# lo cual en realidad, no es de nuestro interés por lo cual,
# no es muy significativo
vif = vif[vif["Variable"] != "const"]

# Ordenar de mayor a menor VIF
vif = vif.sort_values("VIF", ascending=False).reset_index(drop=True)
# Resultado final
vif
```

## **Exploración complementaria: endogeneidad, IV y 2SLS**

El modelo central del proyecto es el modelo log-log estimado por WLS con errores estándar robustos HC3. Sin embargo, como ejercicio complementario, se explora una especificación por variables instrumentales mediante 2SLS.

La motivación econométrica es que el ingreso del hogar puede estar correlacionado con factores no observados que también afectan el gasto monetario total. Por ejemplo, características no observadas de preferencias, estabilidad económica, acceso a crédito o hábitos de consumo pueden influir tanto en el ingreso como en el gasto.

En términos generales, el modelo principal puede escribirse como:

$ \log(Gasto_i) = \beta_0 + \beta_1 \log(Ingresos_i) + \beta_2 X_i + u_i $

donde $X_i$ contiene los controles del modelo principal: rubros de gasto, número de integrantes y variables categóricas del hogar.

La posible endogeneidad se expresa como:

$ Cov(\log(Ingresos_i),u_i) \neq 0 $

Si esta condición se cumple, el estimador MCO o WLS puede estar sesgado. Por ello, se estima un modelo IV / 2SLS como prueba complementaria. Esta alternativa no sustituye al modelo principal, porque requiere supuestos fuertes sobre la validez de los instrumentos.

```python
# Preparación de la base para IV / 2SLS

!pip install linearmodels -q

import numpy as np
import pandas as pd
import statsmodels.api as sm
from linearmodels.iv import IV2SLS

# Usamos data_frame1 porque conserva las columnas originales:
# Sexo, Nivel_Educativo, Estrato y Zona
df_iv = data_frame1.copy()

# Verificación de columnas necesarias
columnas_necesarias = [
    "Gasto", "Ingresos", "Alimentos", "Transporte", "Personales",
    "Limpieza", "Educación", "Vivienda", "Salud", "Integrantes",
    "Edad", "Sexo", "Nivel_Educativo", "Estrato", "Zona"
]

faltantes = [col for col in columnas_necesarias if col not in df_iv.columns]

if len(faltantes) > 0:
    raise ValueError(f"Faltan columnas en data_frame1: {faltantes}")
else:
    print("Todas las columnas necesarias están disponibles.")

# Recalcular logs para asegurar consistencia con la base actual
# Esto es importante porque en el notebook se winsorizaron variables después
# de haber creado algunas transformaciones logarítmicas.
df_iv["log_gasto"] = np.log(df_iv["Gasto"] + 1)
df_iv["log_ingresos"] = np.log(df_iv["Ingresos"] + 1)
df_iv["log_alimentos"] = np.log(df_iv["Alimentos"] + 1)
df_iv["log_transporte"] = np.log(df_iv["Transporte"] + 1)
df_iv["log_personales"] = np.log(df_iv["Personales"] + 1)
df_iv["log_limpieza"] = np.log(df_iv["Limpieza"] + 1)
df_iv["log_educacion"] = np.log(df_iv["Educación"] + 1)
df_iv["log_vivienda"] = np.log(df_iv["Vivienda"] + 1)
df_iv["log_salud"] = np.log(df_iv["Salud"] + 1)

# Instrumentos exploratorios para el ingreso
df_iv["Edad2"] = df_iv["Edad"] ** 2

# Eliminar infinitos y faltantes
df_iv = df_iv.replace([np.inf, -np.inf], np.nan).dropna().copy()

print("Dimensión final de df_iv:", df_iv.shape)
print(df_iv.columns.tolist())
```

## **Elección exploratoria de instrumentos**

Para explorar la posible endogeneidad de $\log(Ingresos)$, se usan como instrumentos la edad del jefe del hogar y su término cuadrático:

$ Z_i = \{Edad_i, Edad_i^2\} $

La idea detrás de esta elección es que la edad del jefe del hogar puede estar relacionada con el ingreso a través del ciclo laboral, experiencia acumulada y etapa del ciclo de vida del hogar. Por ello, puede cumplir parcialmente la condición de relevancia:

$ Cov(Z_i,\log(Ingresos_i)) \neq 0 $

Sin embargo, la condición de exogeneidad es más difícil de defender:

$ Cov(Z_i,u_i)=0 $

La edad también podría afectar directamente los patrones de consumo del hogar. Por esa razón, este modelo se interpreta únicamente como una prueba exploratoria. La literatura de variables instrumentales enfatiza que la validez de un instrumento requiere relevancia y exclusión, y esta segunda condición normalmente depende de una argumentación económica fuerte.

```python
# Construcción de matrices para IV / 2SLS

y_var = "log_gasto"
endog_var = "log_ingresos"

controles_continuos = [
    "log_alimentos",
    "log_transporte",
    "log_personales",
    "log_limpieza",
    "log_educacion",
    "log_vivienda",
    "log_salud",
    "Integrantes"
]

controles_categoricos = [
    "Sexo",
    "Nivel_Educativo",
    "Estrato",
    "Zona"
]

instrumentos_excluidos = [
    "Edad",
    "Edad2"
]

# Dummies de variables categóricas
dummies = pd.get_dummies(
    df_iv[controles_categoricos],
    drop_first=True,
    dtype=float
)

# Variables exógenas estructurales
X_exog = pd.concat(
    [
        df_iv[controles_continuos].astype(float),
        dummies
    ],
    axis=1
)

X_exog = sm.add_constant(X_exog)

# Variable dependiente
y = df_iv[y_var].astype(float)

# Variable potencialmente endógena
X_endog = df_iv[[endog_var]].astype(float)

# Instrumentos excluidos
Z = df_iv[instrumentos_excluidos].astype(float)

print("y:", y.shape)
print("X_exog:", X_exog.shape)
print("X_endog:", X_endog.shape)
print("Z:", Z.shape)
```

```python
# Primera etapa: explicar log_ingresos

X_primera_etapa = pd.concat([X_exog, Z], axis=1)

primera_etapa = sm.OLS(
    X_endog[endog_var],
    X_primera_etapa
).fit(cov_type="HC3")

print(primera_etapa.summary())
```

## **Interpretación de la primera etapa**

La primera etapa del modelo IV / 2SLS evalúa si los instrumentos propuestos, $Edad$ y $Edad^2$, tienen capacidad explicativa sobre la variable potencialmente endógena $\log(Ingresos)$.

La ecuación estimada es:

$ \log(Ingresos_i) = \pi_0 + \pi_1 Edad_i + \pi_2 Edad_i^2 + \pi_3 X_i + v_i $

donde $X_i$ representa el conjunto de controles incluidos en el modelo: rubros de gasto, número de integrantes y variables categóricas asociadas al sexo, nivel educativo, estrato y zona.

En términos globales, la primera etapa presenta un ajuste moderado, con $R^2 = 0.516$ y $R^2$ ajustado de $0.515$. Esto indica que el conjunto completo de variables incluidas en la primera etapa sí tiene capacidad para explicar parte de la variación de $\log(Ingresos)$.

Sin embargo, al revisar específicamente los instrumentos excluidos, se observa que $Edad$ no resulta estadísticamente significativa, con un p-value de $0.503$, y $Edad^2$ tampoco resulta significativa al nivel usual de $5\%$, con un p-value de $0.114$. Por tanto, aunque el modelo de primera etapa tiene un ajuste razonable, la evidencia de relevancia instrumental de $Edad$ y $Edad^2$ es débil.

Esto es importante porque, para que un instrumento sea válido, debe cumplir dos condiciones. La primera es la relevancia:

$ Cov(Z_i,\log(Ingresos_i)) \neq 0 $

La segunda es la exogeneidad o restricción de exclusión:

$ Cov(Z_i,u_i)=0 $

Los resultados sugieren que la primera condición no se cumple con suficiente fuerza para los instrumentos propuestos. Además, la segunda condición también es difícil de defender conceptualmente, ya que la edad del jefe del hogar podría afectar directamente los patrones de consumo y no únicamente el ingreso.

Por esta razón, el modelo IV / 2SLS debe interpretarse como una prueba de sensibilidad y no como sustituto del modelo principal. La primera etapa no proporciona evidencia suficientemente fuerte para preferir la especificación IV sobre el modelo log-log estimado por WLS con errores robustos HC3.

```python
# Modelo IV / 2SLS
# Variable endógena: log_ingresos
# Instrumentos excluidos: Edad y Edad2

modelo_iv_2sls = IV2SLS(
    dependent=y,
    exog=X_exog,
    endog=X_endog,
    instruments=Z
).fit(cov_type="robust")

print(modelo_iv_2sls.summary)
```

```python
# Diagnósticos del modelo IV / 2SLS

print("Primera etapa del modelo IV:")
print(modelo_iv_2sls.first_stage)

print("\nPrueba Wu-Hausman de endogeneidad:")
try:
    print(modelo_iv_2sls.wu_hausman())
except Exception as e:
    print("No se pudo calcular Wu-Hausman:", e)

print("\nPrueba Durbin de endogeneidad:")
try:
    print(modelo_iv_2sls.durbin())
except Exception as e:
    print("No se pudo calcular Durbin:", e)

print("\nPrueba de sobreidentificación de Sargan:")
try:
    print(modelo_iv_2sls.sargan)
except Exception as e:
    print("No se pudo calcular Sargan:", e)
```

```python
# Comparación: OLS-HC3, WLS-HC3 e IV/2SLS

X_ols = pd.concat([X_exog, X_endog], axis=1)

modelo_ols_hc3_comparable = sm.OLS(
    y,
    X_ols
).fit(cov_type="HC3")

# Intentar recuperar los pesos del modelo final
try:
    pesos_iv = pd.Series(np.asarray(pesos), index=data_frame1.index).loc[df_iv.index].astype(float)

    modelo_wls_hc3_comparable = sm.WLS(
        y,
        X_ols,
        weights=pesos_iv
    ).fit(cov_type="HC3")

    wls_disponible = True

except Exception as e:
    print("No fue posible usar pesos del modelo_gls:", e)
    modelo_wls_hc3_comparable = None
    wls_disponible = False

comparacion = pd.DataFrame({
    "Modelo": [
        "OLS-HC3 comparable",
        "WLS-HC3 comparable",
        "IV / 2SLS"
    ],
    "Variable dependiente": [
        "log_gasto",
        "log_gasto",
        "log_gasto"
    ],
    "Tratamiento": [
        "Sin instrumentar",
        "Ponderado con pesos del modelo final",
        "log_ingresos instrumentado con Edad y Edad2"
    ],
    "R2": [
        modelo_ols_hc3_comparable.rsquared,
        modelo_wls_hc3_comparable.rsquared if wls_disponible else np.nan,
        modelo_iv_2sls.rsquared
    ],
    "Coef_log_ingresos": [
        modelo_ols_hc3_comparable.params.get("log_ingresos", np.nan),
        modelo_wls_hc3_comparable.params.get("log_ingresos", np.nan) if wls_disponible else np.nan,
        modelo_iv_2sls.params.get("log_ingresos", np.nan)
    ],
    "Pvalue_log_ingresos": [
        modelo_ols_hc3_comparable.pvalues.get("log_ingresos", np.nan),
        modelo_wls_hc3_comparable.pvalues.get("log_ingresos", np.nan) if wls_disponible else np.nan,
        modelo_iv_2sls.pvalues.get("log_ingresos", np.nan)
    ]
})

comparacion
```

La tabla comparativa no muestra que IV/2SLS supere al modelo final `modelo_gls`. Más bien, muestra que al instrumentar `log_ingresos` dentro de una especificación comparable, los resultados cambian poco y no aparece evidencia fuerte de endogeneidad. Por tanto, no hay una razón estadística clara para abandonar el modelo principal WLS-HC3.

## **Exploración complementaria: sistema de ecuaciones simultáneas**

También se explora un sistema de ecuaciones simultáneas de forma aproximada, estimando cada ecuación mediante 2SLS.

La motivación es que ingreso y gasto pueden determinarse conjuntamente. Por un lado, el ingreso condiciona la capacidad de gasto del hogar. Por otro lado, el gasto observado puede estar relacionado con decisiones económicas del hogar, acceso a crédito, liquidez o características no observadas que también se relacionan con el ingreso.

Una forma simplificada del sistema es:

$ \log(Gasto_i) = \alpha_0 + \alpha_1 \log(Ingresos_i) + \alpha_2 X_i + u_i $

$ \log(Ingresos_i) = \gamma_0 + \gamma_1 \log(Gasto_i) + \gamma_2 W_i + v_i $

Este sistema no se propone como modelo principal. Su objetivo es mostrar que se consideró la posibilidad de simultaneidad, aunque la identificación del sistema requiere instrumentos fuertes y supuestos más exigentes que los del modelo WLS-HC3.

```python
# Sistema exploratorio - Ecuación 1
# log_gasto como variable dependiente
# log_ingresos como variable endógena

sistema_eq_gasto = IV2SLS(
    dependent=y,
    exog=X_exog,
    endog=X_endog,
    instruments=Z
).fit(cov_type="robust")

print(sistema_eq_gasto.summary)
```

```python
# Sistema exploratorio - Ecuación 2
# log_ingresos como variable dependiente
# log_gasto como variable endógena

# Variable dependiente de la segunda ecuación
y_ingreso = df_iv["log_ingresos"].astype(float)

# Variable endógena de la segunda ecuación
endog_gasto = df_iv[["log_gasto"]].astype(float)

# Controles exógenos de la ecuación de ingreso
X_ingreso_exog = pd.concat(
    [
        df_iv[["Integrantes", "Edad", "Edad2"]].astype(float),
        dummies
    ],
    axis=1
)

X_ingreso_exog = sm.add_constant(X_ingreso_exog)

# Instrumentos excluidos para log_gasto
# Son fuertes por construcción, pero su validez económica es limitada.
Z_gasto = df_iv[
    [
        "log_alimentos",
        "log_transporte",
        "log_personales",
        "log_limpieza",
        "log_educacion",
        "log_vivienda",
        "log_salud"
    ]
].astype(float)

sistema_eq_ingreso = IV2SLS(
    dependent=y_ingreso,
    exog=X_ingreso_exog,
    endog=endog_gasto,
    instruments=Z_gasto
).fit(cov_type="robust")

print(sistema_eq_ingreso.summary)
```

## **Conclusión de la exploración IV y de ecuaciones simultáneas**

Los modelos IV / 2SLS y el sistema exploratorio de ecuaciones simultáneas se estimaron con el objetivo de revisar si los resultados del modelo principal son sensibles a posibles problemas de endogeneidad o simultaneidad.

En el modelo IV / 2SLS, $\log(Ingresos)$ se trató como una variable potencialmente endógena y se instrumentó mediante $Edad$ y $Edad^2$. El modelo IV arrojó un coeficiente positivo y estadísticamente significativo para $\log(Ingresos)$, aproximadamente igual a $0.193$. Esto implica que un aumento de $1\%$ en el ingreso se asocia con un aumento aproximado de $0.193\%$ en el gasto monetario, manteniendo constantes las demás variables.

Sin embargo, los diagnósticos no respaldan con fuerza la necesidad de usar IV. Por un lado, en la primera etapa los instrumentos $Edad$ y $Edad^2$ no resultaron estadísticamente significativos de manera individual, lo que sugiere una relevancia instrumental débil. Por otro lado, las pruebas de endogeneidad no rechazan la hipótesis nula de exogeneidad de $\log(Ingresos)$. La prueba Wu-Hausman presenta un p-value de $0.2490$ y la prueba de Durbin un p-value de $0.2484$. Por tanto, no hay evidencia estadística suficiente para afirmar que $\log(Ingresos)$ sea endógena en esta especificación.

La comparación entre modelos también muestra que IV / 2SLS no ofrece una mejora sustancial frente a las especificaciones comparables sin instrumentación. Los modelos OLS-HC3 comparable, WLS-HC3 comparable e IV / 2SLS presentan valores de $R^2$ muy similares, alrededor de $0.832$. Además, el coeficiente de $\log(Ingresos)$ cambia poco al instrumentarlo, pasando de aproximadamente $0.199$ en las especificaciones comparables a $0.193$ en IV / 2SLS. Esto refuerza la idea de que la instrumentación no modifica de manera importante la conclusión económica del modelo.

Respecto al sistema exploratorio de ecuaciones simultáneas, la primera ecuación coincide con el planteamiento IV donde $\log(Gasto)$ depende de $\log(Ingresos)$ instrumentado. En la segunda ecuación, $\log(Ingresos)$ se explica usando $\log(Gasto)$ como variable endógena. No obstante, la identificación de esta segunda ecuación presenta una limitación importante: se utilizan rubros de gasto como instrumentos de $\log(Gasto)$, tales como alimentos, transporte, gastos personales, limpieza, educación, vivienda y salud. Estos rubros forman parte del gasto total o están directamente relacionados con él, por lo que su validez como instrumentos es difícil de defender desde el punto de vista económico. En consecuencia, aunque el sistema simultáneo permite explorar una posible relación bidireccional entre ingreso y gasto, no proporciona una alternativa más sólida que el modelo principal.

En conclusión, los modelos IV / 2SLS y de ecuaciones simultáneas funcionan como ejercicios complementarios de sensibilidad, pero no desplazan al modelo central del proyecto. La evidencia obtenida no muestra una endogeneidad significativa de $\log(Ingresos)$, los instrumentos propuestos presentan debilidad en la primera etapa y el sistema simultáneo requiere supuestos de identificación más fuertes. Por ello, la narrativa principal se mantiene en el modelo log-log estimado por WLS, con pesos derivados de una modelación auxiliar de la varianza y errores estándar robustos HC3, ya que ofrece una especificación más clara, directamente interpretable en términos de elasticidades y más adecuada para los objetivos del proyecto.

## **Justificación teórica de la especificación**

La elección de una especificación log-log se relaciona con la literatura de curvas de Engel, donde se estudia cómo el gasto de los hogares cambia con el ingreso o con el gasto total. En este tipo de modelos, las transformaciones logarítmicas permiten interpretar los coeficientes como elasticidades, lo cual es útil para analizar la sensibilidad del gasto ante cambios porcentuales en el ingreso y en distintos componentes del consumo.

La inclusión del ingreso se justifica porque las curvas de Engel describen la relación entre el gasto de los hogares y su nivel de ingreso. La inclusión de rubros como alimentos, transporte, vivienda, educación, salud, limpieza y gastos personales responde a que el gasto monetario total está compuesto por distintas categorías de consumo. Además, variables demográficas como el número de integrantes, zona, sexo del jefe del hogar, nivel educativo y estrato socioeconómico permiten controlar diferencias estructurales entre hogares.

La exploración con variables instrumentales se justifica porque puede existir endogeneidad si el ingreso está relacionado con factores no observados que también afectan el gasto. En términos econométricos, esto implicaría $Cov(\log(Ingresos_i),u_i)\neq 0$. El método IV / 2SLS busca corregir este problema usando instrumentos relacionados con la variable endógena, pero no relacionados directamente con el término de error.

Aun así, la especificación IV se considera secundaria, ya que la validez de los instrumentos es más difícil de defender que la estrategia principal del proyecto. Por tanto, el modelo WLS-HC3 se conserva como modelo final, mientras que IV y ecuaciones simultáneas se presentan como pruebas de sensibilidad.
